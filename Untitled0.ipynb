```python
import spotipy
from spotipy.oauth2 import SpotifyOAuth
import pandas as pd

CLIENT_ID = "5e7fd38e36da48da99d31c457b8ba5d2"
CLIENT_SECRET = "dd2f8db095d94ab9b387b08785a75851"
REDIRECT_URI = "https://example.com/callback"

sp = spotipy.Spotify(auth_manager=SpotifyOAuth(
    client_id=CLIENT_ID,
    client_secret=CLIENT_SECRET,
    redirect_uri=REDIRECT_URI,
    scope="playlist-read-private",
    open_browser=False
))

track_list = []
limit_per_call = 10

for offset in range(0, 50, limit_per_call):
    results = sp.search(q='year:2024', type='track', limit=limit_per_call, offset=offset)
    tracks = results.get('tracks', {}).get('items', [])

    for track in tracks:
        if not track:
            continue

        # Sikker udtrækning af kunstner og album
        artists = track.get('artists', [])
        artist_name = artists[0].get('name', 'Ukendt') if artists else 'Ukendt'

        album = track.get('album', {})
        album_name = album.get('name', 'Ukendt')
        release_date = album.get('release_date', None)

        duration_ms = track.get('duration_ms', 0)

        track_list.append({
            'track_name': track.get('name', 'Ukendt'),
            'artist_name': artist_name,
            'popularity': track.get('popularity', None),  # Sætter None hvis feltet mangler
            'duration_ms': duration_ms,
            'duration_min': round(duration_ms / (1000 * 60), 2) if duration_ms else 0,
            'explicit': track.get('explicit', False),
            'release_date': release_date,
            'album_name': album_name
        })

df = pd.DataFrame(track_list).drop_duplicates(subset=['track_name', 'artist_name'])

print(f"Succes! Hentet {len(df)} sange.")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Sikr at explicit er tekst
df['explicit_tekst'] = df['explicit'].map({True: 'Ja', False: 'Nej'}).fillna('Nej')

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Graf 1: Fordeling af sanglængde (Histogram)
sns.histplot(df['duration_min'], kde=True, ax=axes[0], color='forestgreen', bins=6)
axes[0].set_title('Fordeling af sanglængde (minutter)')
axes[0].set_xlabel('Minutter')
axes[0].set_ylabel('Antal sange')

# Tjek om popularity indeholder reelle tal
har_popularitet = df['popularity'].dropna().count() > 0

if har_popularitet:
    # Graf 2: Popularitet vs. Varighed
    sns.scatterplot(
        data=df, x='duration_min', y='popularity', hue='explicit_tekst',
        s=100, ax=axes[1], palette={'Ja': '#e74c3c', 'Nej': '#2ecc71'}
    )
    axes[1].set_title('Popularitet vs. Varighed')
    axes[1].set_xlabel('Varighed (minutter)')
    axes[1].set_ylabel('Popularitet (0-100)')

    # Graf 3: Gns. popularitet
    gns = df.groupby('explicit_tekst')['popularity'].mean()
    axes[2].bar(gns.index, gns.values, color=['#2ecc71' if k == 'Nej' else '#e74c3c' for k in gns.index], width=0.4)
    axes[2].set_title('Gns. popularitet: Explicit vs. Clean')
    axes[2].set_ylabel('Popularitet')
    axes[2].set_ylim(0, 100)
    for i, v in enumerate(gns.values):
        axes[2].text(i, v + 2, f"{v:.1f}", ha='center', fontweight='bold')
else:
    # ALTERNATIV HVIS POPULARITET MANGLER:
    # Graf 2: Varighed pr. sang (overblik)
    df_sorted = df.sort_values(by='duration_min', ascending=False)
    axes[1].barh(df_sorted['track_name'][:10], df_sorted['duration_min'][:10], color='skyblue')
    axes[1].set_title('Top 10 længste sange')
    axes[1].set_xlabel('Varighed (minutter)')

    # Graf 3: Gns. varighed for Explicit vs. Clean
    gns_dur = df.groupby('explicit_tekst')['duration_min'].mean()
    axes[2].bar(gns_dur.index, gns_dur.values, color=['#2ecc71' if k == 'Nej' else '#e74c3c' for k in gns_dur.index], width=0.4)
    axes[2].set_title('Gns. varighed: Explicit vs. Clean')
    axes[2].set_ylabel('Minutter')
    for i, v in enumerate(gns_dur.values):
        axes[2].text(i, v + 0.1, f"{v:.2f} min", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
print(df[['track_name', 'popularity', 'duration_min']].head(10))
print("\nEr der tal i popularity?:", df['popularity'].notna().sum(), "ud af", len(df))
```

```python
import pandas as pd

# 1. Hent track IDs i batches af 10 (højst tilladte grænse)
track_ids = []
for offset in [0, 10, 20]:
    results = sp.search(q='year:2024', type='track', limit=10, offset=offset)
    items = results.get('tracks', {}).get('items', [])
    for t in items:
        if t and t.get('id'):
            track_ids.append(t['id'])

# Fjern dubletter blandt ID'erne
track_ids = list(dict.fromkeys(track_ids))

# 2. Hent de fulde metadata via sp.tracks (her følger popularitet med)
full_tracks_response = sp.tracks(track_ids)
full_tracks = full_tracks_response.get('tracks', [])

# 3. Byg DataFrame med reelle data
data = []
for t in full_tracks:
    if not t:
        continue
    data.append({
        'track_name': t.get('name', 'Ukendt'),
        'artist_name': t['artists'][0]['name'] if t.get('artists') else 'Ukendt',
        'popularity': t.get('popularity'),
        'duration_min': round(t.get('duration_ms', 0) / (1000 * 60), 2),
        'explicit_tekst': 'Ja' if t.get('explicit') else 'Nej'
    })

df = pd.DataFrame(data)

# Tjek resultatet
print(f"Hentet {len(df)} sange med fulde metadata.")
print(df[['track_name', 'artist_name', 'popularity', 'duration_min']].head())
print("\nAntal med popularitet:", df['popularity'].notna().sum(), "ud af", len(df))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Indlæs direkte fra open data (over 30.000 sange med alle lydfeatures)
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-01-21/spotify_songs.csv"
df = pd.read_csv(url)

# 2. Rens og omregn varighed til minutter
df = df.dropna(subset=['track_name', 'track_artist', 'track_popularity'])
df['duration_min'] = df['duration_ms'] / (1000 * 60)

# Filtrer outliers fra (f.eks. sange under 1 min eller over 8 min)
df = df[(df['duration_min'] >= 1) & (df['duration_min'] <= 8)]

print(f"Datasæt indlæst! Antal sange: {len(df):,}")
df[['track_name', 'track_artist', 'track_popularity', 'playlist_genre', 'danceability', 'energy', 'valence']].head()
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(20, 5))

# Graf 1: Hvilke genrer scorer højest i gennemsnitlig popularitet?
genre_pop = df.groupby('playlist_genre')['track_popularity'].mean().sort_values(ascending=False)
sns.barplot(x=genre_pop.values, y=genre_pop.index, ax=axes[0], palette='viridis', hue=genre_pop.index, legend=False)
axes[0].set_title('Gennemsnitlig popularitet pr. genre')
axes[0].set_xlabel('Popularitetsscore (0-100)')
axes[0].set_ylabel('Genre')

# Graf 2: Korrelationsmatrix mellem lydparametre og popularitet
features = ['track_popularity', 'danceability', 'energy', 'loudness', 'valence', 'tempo', 'duration_min']
corr = df[features].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1, ax=axes[1])
axes[1].set_title('Hvad hænger sammen med popularitet?')

# Graf 3: Stemning (Valence) vs. Energi fordelt på EDM og R&B
sample = df[df['playlist_genre'].isin(['edm', 'r&b'])].sample(1000, random_state=42)
sns.scatterplot(
    data=sample, x='valence', y='energy', hue='playlist_genre',
    alpha=0.6, ax=axes[2], palette={'edm': '#e74c3c', 'r&b': '#3498db'}
)
axes[2].set_title('Stemning vs. Energi: EDM vs. R&B')
axes[2].set_xlabel('Valence (0 = Melankolsk, 1 = Glad)')
axes[2].set_ylabel('Energy (Intensitet)')

plt.tight_layout()
plt.show()
```
